```python
df["Label"] = df["Label"].str.strip()
```

```python
label_map = {
    "BENIGN": 0,
    "Web Attack \x96 Brute Force": 1,
    "Web Attack \x96 XSS": 1,
    "Web Attack \x96 Sql Injection": 1,
}
```

```python
df["target"] = df["Label"].map(label_map)
```

```python
print(df["target"].value_counts())
print(df["target"].isna().sum())
```

```python
drop_cols = ["Flow ID", "Source IP", "Destination IP", "Timestamp", "Label", "target"]
```

```python
X = df.drop(columns=drop_cols)
y = df["target"]
```

```python
print(X.shape)
print(X.dtypes.value_counts())
```

```python
df.columns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
model = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42)
```

```python
model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
y_proba = model.predict_proba(X_test_scaled)[:, 1]
```

```python
print(classification_report(y_test, y_pred, digits=4))
```

```python
print("ROC AUC:", roc_auc_score(y_test, y_proba))
```

```python
from sklearn.metrics import precision_recall_curve
```

```python
precisions, recalls, thresholds = precision_recall_curve(y_test, y_proba)
```

```python
target_recall = 0.90
mask = recalls[:-1] >= target_recall
best_idx = np.where(mask)[0][-1]
```

```python
print(f"threshold={thresholds[best_idx]:.4f}")
print(f"precision={precisions[best_idx]:.4f}")
print(f"recall={recalls[best_idx]:.4f}")
```
